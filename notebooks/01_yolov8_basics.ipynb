```python
# 2. Load the pretrained YOLOv8 model
from ultralytics import YOLO
model = YOLO("yolov8n.pt")

# 3. Test on a sample image
results = model("https://ultralytics.com/images/bus.jpg")

# 4. Save and display bounding box details
print("Detected Classes:", results[0].boxes.cls)
print("Confidence Scores:", results[0].boxes.conf)
results[0].save("output_bus.jpg")
```
